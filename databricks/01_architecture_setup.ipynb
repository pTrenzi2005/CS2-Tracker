```python
dbutils.widgets.text("catalog_name", "cs2_dev")
catalog_name = dbutils.widgets.get("catalog_name")

dbutils.widgets.text("storage_account", "datalakecstracker")
storage_account = dbutils.widgets.get("storage_account")

abfss_path_bronze = f'abfss://bronze@{storage_account}.dfs.core.windows.net/uc/'
abfss_path_silver = f'abfss://silver@{storage_account}.dfs.core.windows.net/uc/'
abfss_path_gold = f'abfss://gold@{storage_account}.dfs.core.windows.net/uc/'
abfss_path_metastore = f'abfss://metastore@{storage_account}.dfs.core.windows.net/uc/'
abfss_path_bronze_volume = f'abfss://bronze@{storage_account}.dfs.core.windows.net/raw/batch_reference_data/'
```

```python
#Bronze
spark.sql(f"""
    CREATE SCHEMA IF NOT EXISTS {catalog_name}.bronze
    MANAGED LOCATION '{abfss_path_bronze}'
""")

#Silver
spark.sql(f"""
    CREATE SCHEMA IF NOT EXISTS {catalog_name}.silver
    MANAGED LOCATION '{abfss_path_silver}'
""")

#Gold
spark.sql(f"""
    CREATE SCHEMA IF NOT EXISTS {catalog_name}.gold
    MANAGED LOCATION '{abfss_path_gold}'
""")

#Metastore
spark.sql(f"""
    CREATE SCHEMA IF NOT EXISTS {catalog_name}.metastore
    MANAGED LOCATION '{abfss_path_metastore}'
""")

#Volume in bronze
spark.sql(f"""
    CREATE EXTERNAL VOLUME IF NOT EXISTS {catalog_name}.bronze.raw
    LOCATION '{abfss_path_bronze_volume}'
""")

```
